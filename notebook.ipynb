```python
projD = pd.read_csv('projD_ori.csv').set_index('Unnamed: 0')
projD.index.name = 'TF'
```

```python
P = pd.read_csv('P_1-1.csv').set_index('Unnamed: 0')
P.index.name='Cell'
```

```python
import subprocess
from tqdm.notebook import tqdm
for i in tqdm(projD.index):
    process = subprocess.Popen(f'wget "https://www.ncbi.nlm.nih.gov/research/coronavirus-api/export/tsv?text={i}"', stdout = subprocess.PIPE,
        stderr = subprocess.PIPE,
        text = True,
        shell = True)
    std_out, std_err = process.communicate()
```
