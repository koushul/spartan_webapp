```python

from utils.dataloaders import Datacore, make_datacore, severus_pair
```

```python
datacore =  make_datacore()()
```

```python
import itertools
import time
from tqdm.notebook import tqdm
```

```python
pairs = list(itertools.product(
    datacore.name2ds["GSE167118"].total_surface_proteins, 
    datacore.name2ds["GSE167118"].total_transcription_factors))
```

```python
results = []
pbar = tqdm(total=len(pairs))

for a, b in pairs:
    df = severus_pair('CD22', 'NFYA')
    time.sleep(3)
    
    if len(df) > 0:
        results.append((a,b, df))
        
    pbar.update()
    pbar.set_description(f'Hits: {len(results)}')

pbar.close()
```
